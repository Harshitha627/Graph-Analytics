```python

import sys
from collections import deque

def solve():
    input = sys.stdin.buffer.readline
    t = int(input())

    for _ in range(t):
        n, m = map(int, input().split())

        adj = [set() for _ in range(n + 1)]

        for _ in range(m):
            u, v = map(int, input().split())
            adj[u].add(v)
            adj[v].add(u)

        s = int(input())

        dist = [-1] * (n + 1)
        dist[s] = 0

        unvisited = set(range(1, n + 1))
        unvisited.remove(s)

        q = deque([s])

        while q:
            u = q.popleft()

            # Check unvisited vertices in one pass.
            # Missing edges represent village roads.
            for v in list(unvisited):
                if v not in adj[u]:
                    dist[v] = dist[u] + 1
                    q.append(v)
                    unvisited.remove(v)

        sys.stdout.write(
            " ".join(str(dist[i]) for i in range(1, n + 1)
                     if i != s) + "\n"
        )

if __name__ == "__main__":
    solve()
```
